## 1. Project introduction

Our project asks whether we can predict the severity of a traffic accident using information about the weather, roadway, time, and location. The goal is to understand which conditions are linked to more severe accidents and to build a classification model that predicts accident severity.

We plan to use the **US Accidents (2016–2023)** dataset available through Kaggle. According to our proposal, it contains approximately **7.7 million accident records, 46 columns, and observations from 49 U.S. states**. The reported CSV size is about **3.06 GB**, with records from February 2016 to March 2023. The target is `Severity`, on a scale from **1 (lower impact) to 4 (higher impact)**.

We chose this topic because serious crashes have consequences for drivers, passengers, road access, and emergency response. Understanding the relationship between accident severity and environmental or road conditions could help identify patterns worth further investigation. **This is an observational dataset, so correlations will not by themselves establish that a condition causes greater severity.**

Our original hypotheses were: **(H1)** worse weather and low visibility are associated with more severe crashes; **(H2)** night and rush-hour crashes have different severity patterns; and **(H3)** road features such as junctions and traffic signals are related to severity.

## 4. Notebook setup and data loading

**To run:** Download the US Accidents CSV from Kaggle and set `DATA_PATH` below to its location. The filename may be `US_Accidents_March23.csv`. The program samples chunks rather than loading the entire 3 GB dataset into memory at once. The fraction is based on the approximate record count in our proposal, so final sample size can differ slightly.

**Suggested setup:** `pip install pandas numpy matplotlib seaborn scikit-learn`

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, accuracy_score, classification_report, ConfusionMatrixDisplay

sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
DATA_PATH = Path("US_Accidents_March23.csv")  # CHANGE to your actual CSV path
TARGET_SAMPLE = 500_000
ESTIMATED_TOTAL_ROWS = 7_700_000  # Approximation from our proposal
ML_SAMPLE_SIZE = 30_000  # Start small for a fast same-day modeling comparison
print("Ready. Dataset path:", DATA_PATH.resolve())
```

```python
required_cols = [
    "Severity", "Start_Time", "State", "City", "Start_Lat", "Start_Lng",
    "Temperature(F)", "Humidity(%)", "Visibility(mi)", "Precipitation(in)",
    "Wind_Speed(mph)", "Weather_Condition", "Sunrise_Sunset",
    "Junction", "Traffic_Signal", "Crossing", "Stop", "Railway",
    "Give_Way", "Roundabout", "No_Exit", "Traffic_Calming", "Amenity",
    "Bump", "Station", "Turning_Loop"
]
if not DATA_PATH.is_file():
    raise FileNotFoundError(
        f"Cannot find {DATA_PATH}. Download the US Accidents CSV and set DATA_PATH before running."
    )

# Sampling separately within each chunk avoids loading the full CSV into RAM.
# Deterministic across repeated runs for the same row ordering.
rng = np.random.default_rng(RANDOM_STATE)
prob = min(1.0, TARGET_SAMPLE / ESTIMATED_TOTAL_ROWS)
parts = []
rows_read = 0
for chunk in pd.read_csv(
    DATA_PATH, usecols=lambda name: name in required_cols,
    chunksize=100_000, low_memory=False
):
    rows_read += len(chunk)
    mask = rng.random(len(chunk)) < prob
    if mask.any():
        parts.append(chunk.loc[mask].copy())
if not parts:
    raise ValueError("The sampler found no rows. Check your CSV file.")
df = pd.concat(parts, ignore_index=True)
if len(df) > TARGET_SAMPLE:
    df = df.sample(n=TARGET_SAMPLE, random_state=RANDOM_STATE).reset_index(drop=True)

missing_expected = sorted(set(required_cols) - set(df.columns))
if missing_expected:
    print("Warning: these expected columns are missing:", missing_expected)
print(f"Read {rows_read:,} raw rows; selected {len(df):,} rows.")
print("Severity counts before cleaning:")
print(df['Severity'].value_counts(dropna=False).sort_index())
display(df.head(3))
```

```python
# Consistent target, date fields, and explanatory features
raw_sample_rows = len(df)
df["Severity"] = pd.to_numeric(df["Severity"], errors="coerce")
df["Start_Time"] = pd.to_datetime(df["Start_Time"], errors="coerce")
df = df.loc[df["Severity"].isin([1, 2, 3, 4]) & df["Start_Time"].notna()].copy()
df["Severity"] = df["Severity"].astype(int)
df["High_Severity"] = (df["Severity"] >= 3).astype(int)
df["Hour"] = df["Start_Time"].dt.hour
df["Weekday"] = df["Start_Time"].dt.dayofweek  # Monday=0
month = df["Start_Time"].dt.month
df["Season"] = np.select(
    [month.isin([12, 1, 2]), month.isin([3, 4, 5]),
     month.isin([6, 7, 8]), month.isin([9, 10, 11])],
    ["Winter", "Spring", "Summer", "Fall"], default="Unknown"
)
df["Rush_Hour"] = ((df["Weekday"] < 5) &
                  (df["Hour"].between(7, 9) | df["Hour"].between(16, 18))).astype(int)

conditions = df["Weather_Condition"].fillna("Unknown").astype(str).str.lower()
df["Weather_Group"] = np.select([
    conditions.str.contains("snow|sleet|ice|freez", regex=True),
    conditions.str.contains("rain|drizzle|shower|thunder", regex=True),
    conditions.str.contains("fog|mist|haze|smoke", regex=True),
    conditions.str.contains("clear|fair"),
    conditions.str.contains("cloud|overcast")
], ["Snow/ice", "Rain/storm", "Fog/haze", "Clear/fair", "Cloudy"], default="Other/unknown")

road_cols = [col for col in ["Junction", "Traffic_Signal", "Crossing", "Stop",
                            "Railway", "Give_Way", "Roundabout", "No_Exit",
                            "Traffic_Calming", "Amenity", "Bump", "Station",
                            "Turning_Loop"] if col in df.columns]
for col in road_cols:
    if df[col].dtype == bool:
        df[col] = df[col].astype(float)
    else:
        df[col] = df[col].astype(str).str.lower().map(
            {"true": 1.0, "false": 0.0, "1": 1.0, "0": 0.0}
        )
for col in ["Temperature(F)", "Humidity(%)", "Visibility(mi)",
            "Precipitation(in)", "Wind_Speed(mph)"]:
    if col in df:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print(f"Valid rows: {len(df):,} / {raw_sample_rows:,}")
print("Severity distribution (%):")
print((df["Severity"].value_counts(normalize=True).sort_index() * 100).round(2))
print("Missing values in selected fields (%):")
print((df.isna().mean().sort_values(ascending=False).head(12) * 100).round(1))
```

Before comparing conditions, we first inspect the target distribution and missing values. The severity classes may be imbalanced, meaning raw accuracy could overstate predictive performance. We therefore plan to use **macro F1**, which gives each class equal weight when averaging its F1 score, and to examine a confusion matrix.

```python
# Descriptive figure (additional; not counted as one of the six hypotheses)
plt.figure(figsize=(7, 4))
ax = sns.countplot(data=df, x="Severity", order=[1, 2, 3, 4], color="steelblue")
ax.set(title="Distribution of reported accident severity", xlabel="Severity class", ylabel="Sampled accidents")
plt.tight_layout(); plt.show()
```

### Hypothesis chart 1: Visibility and severity
**Hypothesis:** Accidents reported under lower visibility may have a greater share of high-severity cases.  
**Why investigate it?** Low visibility can limit what drivers see; the relationship can be checked directly in accident reports.  
**Responsible member(s): [Assign at team meeting]**  
**[Complete after running: report the group rates/counts, note missing visibility values, and say whether the pattern supports H1.]**

```python
tmp = df.dropna(subset=["Visibility(mi)"]).copy()
tmp = tmp[tmp["Visibility(mi)"] >= 0]
tmp["Visibility_Bin"] = pd.cut(tmp["Visibility(mi)"],
    bins=[-0.001, 1, 3, 10, float("inf")],
    labels=["0–1 mi", ">1–3 mi", ">3–10 mi", ">10 mi"], include_lowest=True)
stats = tmp.groupby("Visibility_Bin", observed=True)["High_Severity"].agg(["mean", "count"]).reset_index()
stats["Percent_High"] = stats["mean"] * 100
print(stats.to_string(index=False))
plt.figure(figsize=(7,4))
sns.barplot(data=stats, x="Visibility_Bin", y="Percent_High", color="steelblue")
plt.title("High severity (3–4) by reported visibility")
plt.ylabel("High-severity accidents (%)"); plt.xlabel("Visibility")
plt.tight_layout(); plt.show()
```

### Hypothesis chart 5: Road features and severity
**Hypothesis:** Crashes at junctions, signals, crossings, and stop-controlled areas may differ in severity.  
**Why investigate it?** Road design and traffic controls change the setting of a crash. This chart compares the high-severity share where each flag is true versus false.  
**Responsible member(s): [Assign at team meeting]**  
**[Complete after running: identify which flags have the largest differences and check group sizes. Flags may overlap, so columns are not mutually exclusive.]**

```python
road_display = [c for c in ["Junction", "Traffic_Signal", "Crossing", "Stop"] if c in df.columns]
records = []
for col in road_display:
    for value, label in [(0.0, "No"), (1.0, "Yes")]:
        subset = df.loc[df[col] == value, "High_Severity"]
        if len(subset):
            records.append({"Feature": col, "Flag": label,
                            "Percent_High": subset.mean() * 100, "Count": len(subset)})
stats = pd.DataFrame(records)
print(stats.to_string(index=False))
plt.figure(figsize=(9,4))
sns.barplot(data=stats, x="Feature", y="Percent_High", hue="Flag")
plt.title("High severity (3–4) and selected roadway flags")
plt.ylabel("High-severity accidents (%)"); plt.xlabel("Road feature")
plt.tight_layout(); plt.show()
```

## 6. Machine-learning analyses

Our proposal calls for predicting the four-level severity label and evaluating models with **macro F1** and a **confusion matrix**. We will compare both models against a **DummyClassifier** baseline that always predicts the most common class. This comparison is important because the classes may be imbalanced, and high overall accuracy could hide poor performance on minority classes.

**Model 1: Logistic Regression** provides a straightforward linear classification benchmark using weather, time, state, and road features.  
**Responsible member(s): [Assign at team meeting]**

**Model 2: Random Forest** captures nonlinear patterns and interactions between the same inputs.  
**Responsible member(s): [Assign at team meeting]**

**Evaluation design:** Hold out 20% of one fixed modeling sample, use a stratified split, and fit preprocessing **only on the training set**. Evaluate the same untouched test rows across baseline and both models. Initial modeling uses at most 30,000 sampled rows to reduce runtime; this is a practical first comparison, **not a claim that the full 500K sample was trained**. If time allows, repeat with more records. We will avoid claiming causation or making deployment claims from these preliminary results.

```python
numeric_candidates = ["Temperature(F)", "Humidity(%)", "Visibility(mi)",
    "Precipitation(in)", "Wind_Speed(mph)", "Hour", "Weekday", "Rush_Hour"]
cat_candidates = ["State", "Weather_Group", "Sunrise_Sunset", "Season"]
numeric_features = [c for c in numeric_candidates + road_cols if c in df.columns]
categorical_features = [c for c in cat_candidates if c in df.columns]
model_cols = numeric_features + categorical_features
model_df = df.dropna(subset=["Severity"]).copy()
# A stratified subset preserves rare severity classes when possible.
if len(model_df) > ML_SAMPLE_SIZE:
    train_model_df, _ = train_test_split(
        model_df, train_size=ML_SAMPLE_SIZE, random_state=RANDOM_STATE,
        stratify=model_df["Severity"]
    )
    model_df = train_model_df
X = model_df[model_cols].copy()
y = model_df["Severity"].astype(int)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
print(f"Training: {len(X_train):,}, test: {len(X_test):,}")
print("Class counts in modeling sample:", y.value_counts().sort_index().to_dict())

# Separate preprocessing pipelines fit to training data only.
numeric_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler())
])
categorical_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer([
    ("num", numeric_pipe, numeric_features),
    ("cat", categorical_pipe, categorical_features)
])
```

```python
# Train baseline and two analyses using the same split and features.
models = {
    "Majority-class baseline": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": Pipeline([
        ("preprocess", preprocessor),
        ("model", LogisticRegression(max_iter=400, class_weight="balanced"))
    ]),
    "Random Forest": Pipeline([
        ("preprocess", preprocessor),
        ("model", RandomForestClassifier(n_estimators=80, max_depth=18,
                      min_samples_leaf=2, random_state=RANDOM_STATE,
                      n_jobs=-1, class_weight="balanced_subsample"))
    ])
}
metrics = []
predictions = {}
for name, clf in models.items():
    print("Training/evaluating:", name)
    clf.fit(X_train, y_train)
    pred = clf.predict(X_test)
    predictions[name] = pred
    metrics.append({"Model": name,
                    "Accuracy": accuracy_score(y_test, pred),
                    "Macro F1": f1_score(y_test, pred, average="macro", zero_division=0),
                    "Weighted F1": f1_score(y_test, pred, average="weighted", zero_division=0)})
results = pd.DataFrame(metrics).sort_values("Macro F1", ascending=False)
display(results.round(3))
```

### Model 2 interpretation: Random Forest
**Responsible member(s): [Assign at team meeting]**  
**[Complete after running:]** State Random Forest's macro F1 and accuracy, compare it against both the majority-class baseline and Logistic Regression, and discuss the confusion matrix and the most influential input features. Feature importance here is descriptive and is **not proof of a causal effect**.

```python
for name in ["Logistic Regression", "Random Forest"]:
    print("\n" + name)
    print(classification_report(y_test, predictions[name], labels=[1, 2, 3, 4], zero_division=0))
    ConfusionMatrixDisplay.from_predictions(
        y_test, predictions[name], labels=[1, 2, 3, 4],
        cmap="Blues", values_format="d"
    )
    plt.title(name + " confusion matrix")
    plt.tight_layout(); plt.show()

rf = models["Random Forest"]
feature_names = rf.named_steps["preprocess"].get_feature_names_out()
importance = pd.Series(rf.named_steps["model"].feature_importances_,
                       index=feature_names).sort_values(ascending=False).head(15)
plt.figure(figsize=(9,5))
importance.sort_values().plot(kind="barh", color="steelblue")
plt.title("Random Forest: top feature importances (predictive, not causal)")
plt.xlabel("Impurity-based importance")
plt.tight_layout(); plt.show()
```
